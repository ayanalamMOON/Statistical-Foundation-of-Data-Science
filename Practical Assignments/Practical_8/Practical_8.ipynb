## 2. Theoretical Background

### 2.1 Decision Tree Fundamentals

#### 2.1.1 What is a Decision Tree?

A **Decision Tree** is a supervised machine learning algorithm that recursively partitions the feature space into regions and makes predictions based on simple decision rules inferred from the data features.

**Tree Structure Components:**

1. **Root Node**: The topmost node representing the entire dataset; contains the feature that best splits the data
2. **Internal Nodes**: Intermediate decision points that split the data based on feature values
3. **Branches**: Connections between nodes representing the outcome of a test/decision
4. **Leaf Nodes (Terminal Nodes)**: Final nodes that contain the prediction/classification

**Mathematical Representation:**

A decision tree can be viewed as a function $f: \mathbb{R}^d \rightarrow \mathcal{Y}$ that maps input features $\mathbf{x} \in \mathbb{R}^d$ to outputs $y \in \mathcal{Y}$.

For classification with $C$ classes, the tree partitions the input space into $M$ disjoint regions $R_1, R_2, ..., R_M$ where:

$$f(\mathbf{x}) = \sum_{m=1}^{M} c_m \cdot \mathbb{1}(\mathbf{x} \in R_m)$$

where $c_m$ is the predicted class for region $R_m$ and $\mathbb{1}(\cdot)$ is the indicator function.

#### 2.1.2 How Decision Trees Make Predictions

**Training Phase:**
1. Start with all training data at the root node
2. Find the feature and split point that best divides the data (maximizes information gain or minimizes impurity)
3. Create child nodes for each partition
4. Recursively repeat steps 2-3 for each child node until stopping criteria are met

**Stopping Criteria:**
- All samples in a node belong to the same class (pure node)
- Maximum tree depth is reached
- Minimum samples required to split a node are not met
- No improvement in impurity measure

**Prediction Phase:**
For a new instance $\mathbf{x}_{new}$:
1. Start at the root node
2. Follow the decision rules based on feature values
3. Traverse down the tree until reaching a leaf node
4. Return the class label (or probability) associated with that leaf

#### 2.1.3 Advantages and Limitations

**Advantages:**
- ✓ **Interpretability**: Easy to understand and explain (white-box model)
- ✓ **No data preprocessing**: Handles both numerical and categorical features
- ✓ **Non-parametric**: No assumptions about data distribution
- ✓ **Feature importance**: Automatically provides feature rankings
- ✓ **Non-linear relationships**: Can capture complex decision boundaries

**Limitations:**
- ✗ **Overfitting**: Can create overly complex trees that don't generalize
- ✗ **Instability**: Small variations in data can result in different trees
- ✗ **Bias**: Biased towards features with more levels
- ✗ **Local optima**: Greedy algorithm may not find globally optimal tree
- ✗ **Class imbalance**: Tends to be biased toward dominant classes

### 2.2 Mathematical Foundations: Entropy

#### 2.2.1 Information Theory Background

**Entropy** is a concept borrowed from information theory, introduced by Claude Shannon in 1948. It measures the **average amount of information** (or uncertainty) in a random variable.

In the context of decision trees, entropy quantifies the **impurity** or **disorder** in a dataset. Higher entropy means more disorder/uncertainty, while lower entropy means more order/predictability.

#### 2.2.2 Mathematical Definition

For a dataset $S$ with $C$ classes, the entropy $H(S)$ is defined as:

$$H(S) = -\sum_{i=1}^{C} p_i \log_2(p_i)$$

where:
- $p_i$ is the proportion of samples belonging to class $i$
- $\log_2$ is the logarithm base 2 (measuring information in bits)
- By convention: $0 \log_2(0) = 0$ (limit as $p \rightarrow 0$)

**For Binary Classification** (our diabetes case with $C=2$):

$$H(S) = -p_0 \log_2(p_0) - p_1 \log_2(p_1)$$

where $p_0$ + $p_1$ = 1

#### 2.2.3 Intuitive Understanding

**Think of entropy as the answer to: "How surprised would I be by a randomly chosen sample's class?"**

- **High Entropy (≈1.0 for binary)**: Classes are approximately balanced → high uncertainty
- **Low Entropy (≈0.0)**: One class dominates → low uncertainty
- **Zero Entropy (=0.0)**: All samples belong to one class → pure node, no uncertainty

**Example Scenarios:**

| Scenario | Class Distribution | Entropy | Interpretation |
|----------|-------------------|---------|----------------|
| Pure Positive | 100% class 1, 0% class 0 | 0.0 bits | Perfect certainty |
| Pure Negative | 0% class 1, 100% class 0 | 0.0 bits | Perfect certainty |
| Balanced | 50% class 1, 50% class 0 | 1.0 bits | Maximum uncertainty |
| Imbalanced | 90% class 1, 10% class 0 | 0.469 bits | Low uncertainty |
| Imbalanced | 70% class 1, 30% class 0 | 0.881 bits | Moderate uncertainty |

#### 2.2.4 Step-by-Step Calculation Example

**Example 1: Perfectly Balanced Dataset**

Suppose we have 100 samples: 50 diabetic (class 1), 50 non-diabetic (class 0)

$$p_1 = \frac{50}{100} = 0.5, \quad p_0 = \frac{50}{100} = 0.5$$

$$H(S) = -0.5 \log_2(0.5) - 0.5 \log_2(0.5)$$
$$H(S) = -0.5 \times (-1) - 0.5 \times (-1)$$
$$H(S) = 0.5 + 0.5 = 1.0 \text{ bits}$$

**Interpretation**: Maximum entropy → Maximum uncertainty → Perfect mix of both classes

**Example 2: Imbalanced Dataset**

Suppose we have 100 samples: 80 diabetic (class 1), 20 non-diabetic (class 0)

$$p_1 = \frac{80}{100} = 0.8, \quad p_0 = \frac{20}{100} = 0.2$$

$$H(S) = -0.8 \log_2(0.8) - 0.2 \log_2(0.2)$$
$$H(S) = -0.8 \times (-0.322) - 0.2 \times (-2.322)$$
$$H(S) = 0.2576 + 0.4644 = 0.722 \text{ bits}$$

**Interpretation**: Lower entropy than balanced case → Class 1 dominates → Less uncertainty

#### 2.2.5 Properties of Entropy

1. **Non-negativity**: $H(S) \geq 0$ always
2. **Maximum for uniform distribution**: For $C$ classes, $H_{max} = \log_2(C)$
3. **Minimum for pure node**: $H_{min} = 0$ when all samples are same class
4. **Concave function**: Entropy is a concave function of the probability distribution
5. **Additive for independent events**: $H(X, Y) = H(X) + H(Y)$ if $X$ and $Y$ are independent

#### 2.2.6 Why Use Entropy in Decision Trees?

Decision trees use entropy to:
1. **Measure node impurity**: Determine how mixed the classes are at each node
2. **Guide splitting decisions**: Choose features that reduce entropy most
3. **Decide when to stop**: Stop splitting when entropy is very low (pure or nearly pure nodes)
4. **Compare split quality**: Higher entropy reduction = better split

### 2.3 Mathematical Foundations: Information Gain

#### 2.3.1 Definition and Purpose

**Information Gain (IG)**, also called **Kullback-Leibler Divergence** or **Mutual Information**, measures the reduction in entropy achieved by splitting the dataset on a particular feature.

**Core Idea**: How much does knowing the value of a feature reduce our uncertainty about the class label?

**Mathematical Definition:**

For a dataset $S$ and a feature $A$:

$$IG(S, A) = H(S) - H(S|A)$$

where:
- $H(S)$ = entropy of the dataset before the split
- $H(S|A)$ = weighted average entropy after splitting on feature $A$

The conditional entropy $H(S|A)$ is calculated as:

$$H(S|A) = \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot H(S_v)$$

where:
- $Values(A)$ = all possible values of feature $A$
- $S_v$ = subset of $S$ where feature $A$ has value $v$
- $|S_v|$ = number of samples in subset $S_v$
- $|S|$ = total number of samples
- $H(S_v)$ = entropy of subset $S_v$

**Complete Formula:**

$$IG(S, A) = H(S) - \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot H(S_v)$$

#### 2.3.2 Intuitive Understanding

Think of Information Gain as:
> "How much more organized is our data after we split it using this feature?"

- **High Information Gain**: The feature creates very pure subsets → Good split
- **Low Information Gain**: The feature doesn't help much in separating classes → Poor split
- **Zero Information Gain**: The feature provides no useful information

**Decision trees select the feature with the highest information gain at each node.**

#### 2.3.3 Step-by-Step Calculation Example

**Example: Predicting Diabetes Based on Age**

Let's say we have 14 patients:

**Original Dataset $S$:**
- Total: 14 patients
- Diabetic (Yes): 9 patients
- Non-Diabetic (No): 5 patients

**Step 1: Calculate Entropy of Original Dataset**

$$p_{yes} = \frac{9}{14} = 0.643, \quad p_{no} = \frac{5}{14} = 0.357$$

$$H(S) = -0.643 \log_2(0.643) - 0.357 \log_2(0.357)$$
$$H(S) = -0.643 \times (-0.638) - 0.357 \times (-1.487)$$
$$H(S) = 0.410 + 0.531 = 0.941 \text{ bits}$$

**Step 2: Split on Feature "Age ≥ 40"**

After splitting on "Age ≥ 40":
- **Subset $S_1$ (Age < 40)**: 5 patients → 2 Yes, 3 No
- **Subset $S_2$ (Age ≥ 40)**: 9 patients → 7 Yes, 2 No

**Step 3: Calculate Entropy for Each Subset**

For $S_1$ (Age < 40):
$$H(S_1) = -\frac{2}{5} \log_2(\frac{2}{5}) - \frac{3}{5} \log_2(\frac{3}{5})$$
$$H(S_1) = -0.4 \times (-1.322) - 0.6 \times (-0.737)$$
$$H(S_1) = 0.529 + 0.442 = 0.971 \text{ bits}$$

For $S_2$ (Age ≥ 40):
$$H(S_2) = -\frac{7}{9} \log_2(\frac{7}{9}) - \frac{2}{9} \log_2(\frac{2}{9})$$
$$H(S_2) = -0.778 \times (-0.364) - 0.222 \times (-2.170)$$
$$H(S_2) = 0.283 + 0.482 = 0.765 \text{ bits}$$

**Step 4: Calculate Weighted Average Entropy**

$$H(S|Age) = \frac{5}{14} \times H(S_1) + \frac{9}{14} \times H(S_2)$$
$$H(S|Age) = \frac{5}{14} \times 0.971 + \frac{9}{14} \times 0.765$$
$$H(S|Age) = 0.347 + 0.492 = 0.839 \text{ bits}$$

**Step 5: Calculate Information Gain**

$$IG(S, Age) = H(S) - H(S|Age)$$
$$IG(S, Age) = 0.941 - 0.839 = 0.102 \text{ bits}$$

**Interpretation**: Splitting on "Age ≥ 40" reduces entropy by 0.102 bits, meaning we gain a small amount of information about diabetes status from this split.

#### 2.3.4 Information Gain for Continuous Features

For continuous features (like Glucose, BMI), we need to discretize:

1. **Sort** the feature values
2. **Consider split points** between consecutive values (or use percentiles)
3. **Calculate IG** for each potential split point
4. **Choose** the split point with maximum IG

**Example**: For Glucose levels [85, 110, 120, 140, 160]:
- Try splits at: 97.5, 115, 130, 150
- Calculate IG for each
- Use the split with highest IG

#### 2.3.5 Properties of Information Gain

1. **Always non-negative**: $IG(S, A) \geq 0$
2. **Maximum value**: $IG(S, A) \leq H(S)$ (can't reduce entropy below 0)
3. **Symmetric**: Order of splits doesn't matter for final IG
4. **Additive**: Can decompose into multiple splits
5. **Bias towards multi-valued attributes**: Features with more distinct values tend to have higher IG (addressed by Gain Ratio)

#### 2.3.6 The ID3 Algorithm

The **ID3 (Iterative Dichotomiser 3)** algorithm uses Information Gain:

```
Algorithm ID3(Dataset S, Features F):
    1. If all samples in S have same class: return leaf node with that class
    2. If F is empty: return leaf node with majority class
    3. Calculate IG(S, A) for each feature A in F
    4. Select feature A* with maximum IG
    5. Create decision node for A*
    6. For each value v of A*:
        - Create subset S_v where A* = v
        - Recursively call ID3(S_v, F - {A*})
    7. Return decision tree
```

### 2.4 Mathematical Foundations: Gini Index

#### 2.4.1 Definition and Origin

The **Gini Index** (or **Gini Impurity**) is an alternative measure of impurity used in decision trees, particularly in the **CART (Classification and Regression Trees)** algorithm.

Named after Italian statistician Corrado Gini (who developed it for measuring income inequality), the Gini Index measures the probability of incorrectly classifying a randomly chosen element if it were randomly labeled according to the class distribution in the dataset.

**Mathematical Definition:**

For a dataset $S$ with $C$ classes:

$$Gini(S) = 1 - \sum_{i=1}^{C} p_i^2$$

where $p_i$ is the proportion of samples belonging to class $i$.

**Alternative (Equivalent) Formulation:**

$$Gini(S) = \sum_{i=1}^{C} p_i(1 - p_i) = \sum_{i=1}^{C} p_i \sum_{j \neq i} p_j$$

This represents the probability of picking a sample of class $i$ times the probability of misclassifying it.

**For Binary Classification:**

$$Gini(S) = 1 - p_0^2 - p_1^2 = 2p_0p_1$$

where $p_0 + p_1 = 1$

#### 2.4.2 Intuitive Understanding

The Gini Index asks: **"If we randomly pick a sample and randomly assign it a label based on the class distribution, what's the probability we're wrong?"**

- **Gini = 0**: Pure node (all samples same class) → 0% chance of error
- **Gini = 0.5**: Balanced  binary classes (50-50) → 50% chance of error (maximum for binary)
- **Higher Gini**: More impurity → Higher chance of misclassification

**Range:**
- Minimum: 0 (pure node)
- Maximum: $1 - \frac{1}{C}$ (uniform distribution over $C$ classes)
- For binary classification: Maximum = 0.5

#### 2.4.3 Step-by-Step Calculation Example

**Example 1: Perfectly Balanced Dataset** (Same as entropy example)

100 samples: 50 diabetic (class 1), 50 non-diabetic (class 0)

$$p_1 = 0.5, \quad p_0 = 0.5$$

$$Gini(S) = 1 - p_0^2 - p_1^2$$
$$Gini(S) = 1 - (0.5)^2 - (0.5)^2$$
$$Gini(S) = 1 - 0.25 - 0.25 = 0.5$$

**Interpretation**: Maximum impurity for binary classification → 50% error probability if randomly labeling

**Example 2: Imbalanced Dataset**

100 samples: 80 diabetic (class 1), 20 non-diabetic (class 0)

$$p_1 = 0.8, \quad p_0 = 0.2$$

$$Gini(S) = 1 - (0.8)^2 - (0.2)^2$$
$$Gini(S) = 1 - 0.64 - 0.04 = 0.32$$

**Interpretation**: Lower impurity than balanced case → 32% error probability → More predictable

**Example 3: Pure Node**

100 samples: 100 diabetic (class 1), 0 non-diabetic (class 0)

$$p_1 = 1.0, \quad p_0 = 0.0$$

$$Gini(S) = 1 - (1.0)^2 - (0.0)^2 = 0$$

**Interpretation**: Perfect purity → 0% error probability → Completely predictable

#### 2.4.4 Gini Index for Splits

Similar to Information Gain, we calculate Gini Gain when splitting on a feature.

**Gini Split (Weighted Gini after split):**

$$Gini_{split}(S, A) = \sum_{v \in Values(A)} \frac{|S_v|}{|S|} \cdot Gini(S_v)$$

**Gini Gain (Reduction in Gini):**

$$\Delta Gini(S, A) = Gini(S) - Gini_{split}(S, A)$$

**Decision trees using CART select the feature that maximizes Gini Gain (or equivalently, minimizes Gini split).**

#### 2.4.5 Worked Example: Gini for a Split

Using our previous diabetes/age example:

**Original Dataset:**
- 14 patients: 9 diabetic, 5 non-diabetic

**Step 1: Calculate Original Gini**

$$Gini(S) = 1 - (\frac{9}{14})^2 - (\frac{5}{14})^2$$
$$Gini(S) = 1 - (0.643)^2 - (0.357)^2$$
$$Gini(S) = 1 - 0.413 - 0.127 = 0.460$$

**Step 2: Split on "Age ≥ 40"**

- $S_1$ (Age < 40): 5 patients → 2 Yes, 3 No
- $S_2$ (Age ≥ 40): 9 patients → 7 Yes, 2 No

**Step 3: Calculate Gini for Each Subset**

For $S_1$:
$$Gini(S_1) = 1 - (\frac{2}{5})^2 - (\frac{3}{5})^2 = 1 - 0.16 - 0.36 = 0.48$$

For $S_2$:
$$Gini(S_2) = 1 - (\frac{7}{9})^2 - (\frac{2}{9})^2 = 1 - 0.605 - 0.049 = 0.346$$

**Step 4: Calculate Weighted Gini**

$$Gini_{split} = \frac{5}{14} \times 0.48 + \frac{9}{14} \times 0.346$$
$$Gini_{split} = 0.171 + 0.223 = 0.394$$

**Step 5: Calculate Gini Gain**

$$\Delta Gini = 0.460 - 0.394 = 0.066$$

**Interpretation**: Splitting on age reduces Gini impurity by 0.066 → Moderately useful split

#### 2.4.6 Entropy vs Gini Index: Detailed Comparison

| Aspect | Entropy | Gini Index |
|--------|---------|------------|
| **Formula** | $-\sum p_i \log_2(p_i)$ | $1 - \sum p_i^2$ |
| **Origin** | Information Theory | Statistics / Economics |
| **Range (binary)** | [0, 1] bits | [0, 0.5] |
| **Maximum (binary)** | 1.0 | 0.5 |
| **Computation** | Requires logarithm | Simple arithmetic |
| **Algorithm** | ID3, C4.5 | CART |
| **Sensitivity** | More sensitive to changes | Less sensitive to changes |
| **Speed** | Slightly slower | Slightly faster |
| **Interpretation** | Information/uncertainty | Misclassification probability |
| **Shape** | Logarithmic | Quadratic |

**Visual Comparison for Binary Classification:**

For probability $p$ of class 1:
- Entropy: $H(p) = -p\log_2(p) - (1-p)\log_2(1-p)$
- Gini: $G(p) = 2p(1-p) = 1 - p^2 - (1-p)^2$

Both are maximized at $p = 0.5$ and minimized at $p = 0$ or $p = 1$.

**Numerical Comparison:**

| $p$ (class 1) | Entropy (bits) | Gini | Normalized Gini |
|--------------|----------------|------|-----------------|
| 0.0 | 0.000 | 0.000 | 0.000 |
| 0.1 | 0.469 | 0.180 | 0.360 |
| 0.2 | 0.722 | 0.320 | 0.640 |
| 0.3 | 0.881 | 0.420 | 0.840 |
| 0.4 | 0.971 | 0.480 | 0.960 |
| 0.5 | 1.000 | 0.500 | 1.000 |

*(Normalized Gini = Gini / 0.5 for comparison)*

**When to use which?**
- **Entropy**: When interpretation as information is important; slightly better theoretical foundation
- **Gini**: When computational efficiency matters; easier to compute; works well in practice
- **In Practice**: Both usually lead to very similar trees; Gini is more common in modern implementations (e.g., `sklearn` defaults to Gini)

#### 2.4.7 Other Impurity Measures

For completeness, other measures exist:
- **Classification Error**: $1 - \max_i(p_i)$ (rarely used, too coarse)
- **Gain Ratio**: Information Gain normalized by split information (C4.5 algorithm)
- **Chi-Square**: Used for feature selection, not tree building

## 5. Feature Selection and Data Preparation

### 5.1 Understanding Feature Selection

**Feature Selection** is the process of identifying which variables (features) are most relevant for making predictions. In supervised learning, we distinguish between:

1. **Independent Variables (Features / Predictors)**: $X$ - Input variables used to make predictions
2. **Dependent Variable (Target / Response)**: $y$ - Output variable we want to predict

**Why Feature Selection Matters:**
- Reduces overfitting by removing irrelevant features
- Improves model performance and accuracy
- Reduces training time
- Enhances model interpretability

**Methods for Feature Selection:**
1. **Domain knowledge**: Use medical/expert knowledge (our approach)
2. **Correlation analysis**: Remove highly correlated features
3. **Recursive Feature Elimination (RFE)**: Iteratively remove least important features
4. **Feature importance**: Use model-based rankings
5. **Regularization**: Use L1/L2 penalties (Lasso, Ridge)

### 5.2 Defining Features and Target

For our diabetes prediction task:
- **Target (y)**: `Outcome` (0 = No Diabetes, 1 = Diabetes)
- **Features (X)**: All other columns (Pregnancies, Glucose, BloodPressure, etc.)

## 6. Data Splitting

### 6.1 Why Split Data?

We split our dataset into **training** and **testing** sets for several critical reasons:

1. **Prevent Overfitting**: Model evaluation on unseen data
2. **Honest Performance Estimate**: Unbiased accuracy assessment
3. **Generalization Assessment**: How well model predicts new cases
4. **Model Selection**: Compare different models fairly

**The Fundamental Rule**: ***Never evaluate your model on the same data used for training!***

### 6.2 Common Split Ratios

| Split Ratio | Training Set | Testing Set | When to Use |
|-------------|--------------|-------------|-------------|
| **80-20** | 80% | 20% | Larger datasets (>1000 samples) |
| **70-30** | 70% | 30% | Medium datasets (500-1000 samples) |
| **60-40** | 60% | 40% | Smaller datasets (<500 samples) |

Our dataset has 768 samples, so we'll use **70-30 split**.

### 6.3 Stratified Splitting

**Stratification** ensures that the class distribution in the training and testing sets matches the original distribution. This is crucial for imbalanced datasets!

**Without Stratification**: Random split might create imbalanced train/test sets
**With Stratification**: Both sets maintain the original class proportions

**Formula for Stratification:**
$$\text{Proportion}_{class_i} = \frac{\text{Count}_{class_i}}{\text{Total Count}}$$

This proportion is preserved in both train and test sets.

### 6.4 Implementing the Split


````python
# Set random seed for reproducibility
RANDOM_STATE = 42

# Display original distribution
print("="*70)
print("ORIGINAL DATASET DISTRIBUTION")
print("="*70)
original_dist = y.value_counts(normalize=True).sort_index() * 100
print("Class 0 (No Diabetes):", f"{original_dist[0]:.2f}%")
print("Class 1 (Diabetes):", f"{original_dist[1]:.2f}%")

# Perform stratified train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,  # 30% for testing
    random_state=RANDOM_STATE,
    stratify=y  # Maintain class distribution
)

print("\n" + "="*70)
print("DATA SPLIT SUMMARY")
print("="*70)
print(f"Total samples: {len(X)}")
print(f"\nTraining set:")
print(f"  Samples: {len(X_train)} ({len(X_train)/len(X)*100:.1f}%)")
print(f"  Features: {X_train.shape[1]}")

print(f"\nTesting set:")
print(f"  Samples: {len(X_test)} ({len(X_test)/len(X)*100:.1f}%)")
print(f"  Features: {X_test.shape[1]}")

# Verify stratification
print("\n" + "="*70)
print("STRATIFICATION VERIFICATION")
print("="*70)

train_dist = y_train.value_counts(normalize=True).sort_index() * 100
test_dist = y_test.value_counts(normalize=True).sort_index() * 100

print("\nClass Distribution Comparison:")
print(f"{'Class':<15} {'Original':<12} {'Training':<12} {'Testing':<12}")
print("-" * 51)
print(f"{'0 (No Diabetes)':<15} {original_dist[0]:>10.2f}% {train_dist[0]:>10.2f}% {test_dist[0]:>10.2f}%")
print(f"{'1 (Diabetes)':<15} {original_dist[1]:>10.2f}% {train_dist[1]:>10.2f}% {test_dist[1]:>10.2f}%")

# Check if stratification worked well
train_diff = abs(train_dist[0] - original_dist[0])
test_diff = abs(test_dist[0] - original_dist[0])

if train_diff < 2 and test_diff < 2:
    print("\n✓ Stratification successful! Distributions match within 2%")
else:
    print("\n⚠ Warning: Stratification shows some deviation")

print("\n" + "="*70)
print("✓ Data splitting complete!")
print("="*70)
````

````python
# Visualize the data split
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Original distribution
axes[0].bar([0, 1], y.value_counts().sort_index().values, color=['#2ecc71', '#e74c3c'], alpha=0.7)
axes[0].set_title('Original Dataset\nN = ' + str(len(y)), fontweight='bold', fontsize=12)
axes[0].set_xlabel('Outcome')
axes[0].set_ylabel('Count')
axes[0].set_xticks([0, 1])
axes[0].set_xticklabels(['No Diabetes', 'Diabetes'])
axes[0].grid(axis='y', alpha=0.3)

# Training set distribution
axes[1].bar([0, 1], y_train.value_counts().sort_index().values, color=['#2ecc71', '#e74c3c'], alpha=0.7)
axes[1].set_title('Training Set (70%)\nN = ' + str(len(y_train)), fontweight='bold', fontsize=12)
axes[1].set_xlabel('Outcome')
axes[1].set_ylabel('Count')
axes[1].set_xticks([0, 1])
axes[1].set_xticklabels(['No Diabetes', 'Diabetes'])
axes[1].grid(axis='y', alpha=0.3)

# Testing set distribution
axes[2].bar([0, 1], y_test.value_counts().sort_index().values, color=['#2ecc71', '#e74c3c'], alpha=0.7)
axes[2].set_title('Testing Set (30%)\nN = ' + str(len(y_test)), fontweight='bold', fontsize=12)
axes[2].set_xlabel('Outcome')
axes[2].set_ylabel('Count')
axes[2].set_xticks([0, 1])
axes[2].set_xticklabels(['No Diabetes', 'Diabetes'])
axes[2].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("✓ Data split visualization complete")
````

## 7. Decision Tree Implementation

### 7.1 DecisionTreeClassifier Parameters

The `sklearn.tree.DecisionTree Classifier` has several important hyperparameters:

| Parameter | Description | Default | Impact |
|-----------|-------------|---------|--------|
| **criterion** | Splitting criterion ('gini' or 'entropy') | 'gini' | How to measure split quality |
| **max_depth** | Maximum depth of tree | None | Deeper = more complex, risk overfitting |
| **min_samples_split** | Minimum samples required to split | 2 | Higher = less splitting, simpler tree |
| **min_samples_leaf** | Minimum samples in a leaf node | 1 | Higher = smoother decision boundaries |
| **random_state** | Seed for reproducibility | None | Ensures consistent results |
| **max_features** | Number of features for best split | None | Can reduce overfitting |

We'll start with default parameters using **Gini Index** as the criterion.

### 7.2 Building the Model


````python
# Initialize the Decision Tree Classifier
print("="*70)
print("DECISION TREE CLASSIFIER - INITIALIZATION")
print("="*70)

# Create model with Gini criterion
dt_classifier = DecisionTreeClassifier(
    criterion='gini',      # Use Gini Index
    random_state=RANDOM_STATE,  # For reproducibility
    max_depth=None,        # No limit on depth (we'll prune later if needed)
    min_samples_split=2,   # Default
    min_samples_leaf=1     # Default
)

print("Model Parameters:")
print(f"  Criterion: {dt_classifier.criterion}")
print(f"  Max Depth: {dt_classifier.max_depth}")
print(f"  Min Samples Split: {dt_classifier.min_samples_split}")
print(f"  Min Samples Leaf: {dt_classifier.min_samples_leaf}")
print(f"  Random State: {dt_classifier.random_state}")

# Train the model
print("\n" + "="*70)
print("TRAINING THE MODEL")
print("="*70)

dt_classifier.fit(X_train, y_train)

print("✓ Model training complete!")
print(f"\nTree Statistics:")
print(f"  Number of leaves: {dt_classifier.get_n_leaves()}")
print(f"  Tree depth: {dt_classifier.get_depth()}")
print(f"  Number of features used: {dt_classifier.n_features_in_}")
````

## 9. Root Node Selection Analysis with Manual Calculations ⭐

### 9.1 Importance of Root Node Selection

The **root node** is the most critical decision in the tree because:
1. It affects ALL subsequent splits
2. The feature chosen creates the first major separation
3. It influences the entire tree structure
4. Poor root = poor overall performance

**Question**: How does the decision tree choose which feature to use at the root?

**Answer**: It selects the feature that provides **maximum information gain** (for entropy criterion) or **maximum Gini gain** (for Gini criterion).

### 9.2 Manual Calculation Strategy

We'll manually calculate:
1. **Entropy of root node** (original dataset entropy)
2. **Information Gain for each feature** (top features)
3. **Gini Index of root node**
4. **Gini Gain for each feature** (top features)
5. **Verify against sklearn's choice**

Since we have continuous features, we'll:
- Use the same split points that sklearn uses
- Or use median splits for demonstration
- Compare results with sklearn's internal calculations

### 9.3 Extract Tree's Root Node Information

### 9.4 Manual Entropy Calculation for Root Node

Let's calculate the entropy of the training dataset before any splits:

**Formula**: $H(S) = -\sum_{i=1}^{C} p_i \log_2(p_i)$

For binary classification:
$H(S) = -p_0 \log_2(p_0) - p_1 \log_2(p_1)$


````python
# Manual entropy calculation for root
print("="*70)
print("MANUAL ENTROPY CALCULATION - ROOT NODE")
print("="*70)

# Class counts in training set
class_counts = y_train.value_counts().sort_index()
n_total = len(y_train)
n_class_0 = class_counts[0]
n_class_1 = class_counts[1]

print(f"Training Set Composition:")
print(f"  Total samples: {n_total}")
print(f"  Class 0 (No Diabetes): {n_class_0}")
print(f"  Class 1 (Diabetes): {n_class_1}")

# Calculate probabilities
p_0 = n_class_0 / n_total
p_1 = n_class_1 / n_total

print(f"\nClass Probabilities:")
print(f"  p(Class 0) = {n_class_0}/{n_total} = {p_0:.6f}")
print(f"  p(Class 1) = {n_class_1}/{n_total} = {p_1:.6f}")

# Calculate entropy manually
# Note: use np.log2 for base 2 logarithm
# Handle 0 * log(0) = 0 case
if p_0 > 0:
    entropy_term_0 = -p_0 * np.log2(p_0)
else:
    entropy_term_0 = 0

if p_1 > 0:
    entropy_term_1 = -p_1 * np.log2(p_1)
else:
    entropy_term_1 = 0

entropy_root = entropy_term_0 + entropy_term_1

print(f"\nEntropy Calculation:")
print(f"  H(S) = -p_0 × log₂(p_0) - p_1 × log₂(p_1)")
print(f"  H(S) = -({p_0:.6f}) × log₂({p_0:.6f}) - ({p_1:.6f}) × log₂({p_1:.6f})")
print(f"  H(S) = -({p_0:.6f}) × ({np.log2(p_0):.6f}) - ({p_1:.6f}) × ({np.log2(p_1):.6f})")
print(f"  H(S) = {entropy_term_0:.6f} + {entropy_term_1:.6f}")
print(f"  H(S) = {entropy_root:.6f} bits")

print(f"\n✓ Root Node Entropy: {entropy_root:.6f} bits")
````

### 9.5 Information Gain Calculation for Top Features

Now we'll calculate Information Gain for each feature to see which one provides the best split.

**Formula**: $IG(S, A) = H(S) - H(S|A)$

where $H(S|A) = \sum_{v} \frac{|S_v|}{|S|} H(S_v)$

We'll calculate this for the actual splits used by sklearn.


````python
# Function to calculate entropy
def calculate_entropy(class_counts):
    """Calculate entropy given class counts."""
    total = sum(class_counts)
    if total == 0:
        return 0.0

 entropy = 0.0
    for count in class_counts:
        if count > 0:
            p = count / total
            entropy -= p * np.log2(p)
    return entropy

# Function to calculate information gain for a feature and threshold
def calculate_information_gain(X, y, feature_idx, threshold):
    """Calculate information gain for a specific split."""
    # Split data
    left_mask = X.iloc[:, feature_idx] <= threshold
    right_mask = ~left_mask

    # Get class counts for each split
    left_classes = y[left_mask].value_counts().sort_index()
    right_classes = y[right_mask].value_counts().sort_index()

    # Ensure both classes are represented (fill with 0 if missing)
    left_counts = [left_classes.get(0, 0), left_classes.get(1, 0)]
    right_counts = [right_classes.get(0, 0), right_classes.get(1, 0)]

    # Calculate entropies
    n_total = len(y)
    n_left = sum(left_counts)
    n_right = sum(right_counts)

    h_left = calculate_entropy(left_counts)
    h_right = calculate_entropy(right_counts)

    # Weighted average entropy after split
    h_split = (n_left / n_total) * h_left + (n_right / n_total) * h_right

    # Information gain
    ig = entropy_root - h_split

    return ig, h_left, h_right, n_left, n_right

print("="*70)
print("INFORMATION GAIN CALCULATIONS FOR ALL FEATURES")
print("="*70)
print("(Using sklearn's chosen split thresholds when available)\n")

# Calculate IG for all features
ig_results = []

for i, feature_name in enumerate(feature_columns):
    # Use the threshold from sklearn if this feature is used in the tree
    # Otherwise, use median as threshold
    feature_used_indices = np.where(tree.feature == i)[0]

    if len(feature_used_indices) > 0:
        # Feature is used in tree, get its threshold from first occurrence
        threshold = tree.threshold[feature_used_indices[0]]
    else:
        # Feature not used, use median
        threshold = X_train.iloc[:, i].median()

    ig, h_left, h_right, n_left, n_right = calculate_information_gain(
        X_train, y_train, i, threshold
    )

    ig_results.append({
        'Feature': feature_name,
        'Threshold': threshold,
        'IG': ig,
        'H_Left': h_left,
        'H_Right': h_right,
        'N_Left': n_left,
        'N_Right': n_right
    })

    # Print detailed calculation for top features or root
    if i == root_feature_idx or ig > 0.05:  # Root or significant IG
        marker = " ⭐ (ROOT NODE)" if i == root_feature_idx else ""
        print(f"\nFeature: {feature_name}{marker}")
        print(f"  Split threshold: {threshold:.4f}")
        print(f"  Left  (≤{threshold:.4f}): {n_left:4d} samples, Entropy = {h_left:.6f}")
        print(f"  Right (>{threshold:.4f}): {n_right:4d} samples, Entropy = {h_right:.6f}")

        weighted_entropy = (n_left/len(y_train)) * h_left + (n_right/len(y_train)) * h_right
        print(f"  Weighted Entropy after split: {weighted_entropy:.6f}")
        print(f"  Information Gain: {entropy_root:.6f} - {weighted_entropy:.6f} = {ig:.6f}")

# Sort by Information Gain
ig_df = pd.DataFrame(ig_results).sort_values('IG', ascending=False)

print("\n" + "="*70)
print("INFORMATION GAIN SUMMARY (Sorted by IG)")
print("="*70)
print(ig_df[['Feature', 'Threshold', 'IG']].to_string(index=False))

print("\n" + "="*70)
print(f"✓ Highest Information Gain: {ig_df.iloc[0]['Feature']} (IG = {ig_df.iloc[0]['IG']:.6f})")
print("="*70)
````

### 9.6 Manual Gini Index Calculation for Root Node

Now let's calculate the Gini Index for the root node:

**Formula**: $Gini(S) = 1 - \sum_{i=1}^{C} p_i^2$

For binary classification:
$Gini(S) = 1 - p_0^2 - p_1^2$


````python
# Manual Gini calculation for root
print("=" *70)
print("MANUAL GINI INDEX CALCULATION - ROOT NODE")
print("="*70)

print(f"Using same class distribution as entropy calculation:")
print(f"  p(Class 0) = {p_0:.6f}")
print(f"  p(Class 1) = {p_1:.6f}")

# Calculate Gini Index manually
gini_root = 1 - p_0**2 - p_1**2

print(f"\nGini Index Calculation:")
print(f"  Gini(S) = 1 - p_0² - p_1²")
print(f"  Gini(S) = 1 - ({p_0:.6f})² - ({p_1:.6f})²")
print(f"  Gini(S) = 1 - {p_0**2:.6f} - {p_1**2:.6f}")
print(f"  Gini(S) = {gini_root:.6f}")

print(f"\n✓ Root Node Gini Index: {gini_root:.6f}")

# Verify against sklearn's reported value
print(f"\nVerification: sklearn's Root Gini = {root_impurity:.6f}")
difference = abs(gini_root - root_impurity)
if difference < 0.0001:
    print("✓ Our manual calculation MATCHES sklearn's value!")
else:
    print(f"⚠ Small difference: {difference:.6f} (likely due to rounding)")
````

### 9.7 Gini Gain Calculation for Top Features

Similar to Information Gain, we calculate Gini Gain for each feature:

**Formula**: $\Delta Gini(S, A) = Gini(S) - Gini_{split}(S, A)$

where $Gini_{split}(S, A) = \sum_{v} \frac{|S_v|}{|S|} Gini(S_v)$


````python
# Function to calculate Gini index
def calculate_gini(class_counts):
    """Calculate Gini index given class counts."""
    total = sum(class_counts)
    if total == 0:
        return 0.0

    gini = 1.0
    for count in class_counts:
        p = count / total
        gini -= p**2
    return gini

# Function to calculate Gini gain for a feature and threshold
def calculate_gini_gain(X, y, feature_idx, threshold):
    """Calculate Gini gain for a specific split."""
    # Split data
    left_mask = X.iloc[:, feature_idx] <= threshold
    right_mask = ~left_mask

    # Get class counts for each split
    left_classes = y[left_mask].value_counts().sort_index()
    right_classes = y[right_mask].value_counts().sort_index()

    # Ensure both classes are represented
    left_counts = [left_classes.get(0, 0), left_classes.get(1, 0)]
    right_counts = [right_classes.get(0, 0), right_classes.get(1, 0)]

    # Calculate Gini indices
    n_total = len(y)
    n_left = sum(left_counts)
    n_right = sum(right_counts)

    gini_left = calculate_gini(left_counts)
    gini_right = calculate_gini(right_counts)

    # Weighted average Gini after split
    gini_split = (n_left / n_total) * gini_left + (n_right / n_total) * gini_right

    # Gini gain
    gini_gain = gini_root - gini_split

    return gini_gain, gini_left, gini_right, n_left, n_right

print("="*70)
print("GINI GAIN CALCULATIONS FOR ALL FEATURES")
print("="*70)
print("(Using sklearn's chosen split thresholds when available)\n")

# Calculate Gini Gain for all features
gini_results = []

for i, feature_name in enumerate(feature_columns):
    # Use threshold from sklearn if available
    feature_used_indices = np.where(tree.feature == i)[0]

    if len(feature_used_indices) > 0:
        threshold = tree.threshold[feature_used_indices[0]]
    else:
        threshold = X_train.iloc[:, i].median()

    gini_gain, g_left, g_right, n_left, n_right = calculate_gini_gain(
        X_train, y_train, i, threshold
    )

    gini_results.append({
        'Feature': feature_name,
        'Threshold': threshold,
        'Gini_Gain': gini_gain,
        'Gini_Left': g_left,
        'Gini_Right': g_right,
        'N_Left': n_left,
        'N_Right': n_right
    })

    # Print detailed calculation for top features or root
    if i == root_feature_idx or gini_gain > 0.02:  # Root or significant gain
        marker = " ⭐ (ROOT NODE)" if i == root_feature_idx else ""
        print(f"\nFeature: {feature_name}{marker}")
        print(f"  Split threshold: {threshold:.4f}")
        print(f"  Left  (≤{threshold:.4f}): {n_left:4d} samples, Gini = {g_left:.6f}")
        print(f"  Right (>{threshold:.4f}): {n_right:4d} samples, Gini = {g_right:.6f}")

        weighted_gini = (n_left/len(y_train)) * g_left + (n_right/len(y_train)) * g_right
        print(f"  Weighted Gini after split: {weighted_gini:.6f}")
        print(f"  Gini Gain: {gini_root:.6f} - {weighted_gini:.6f} = {gini_gain:.6f}")

# Sort by Gini Gain
gini_df = pd.DataFrame(gini_results).sort_values('Gini_Gain', ascending=False)

print("\n" + "="*70)
print("GINI GAIN SUMMARY (Sorted by Gini Gain)")
print("="*70)
print(gini_df[['Feature', 'Threshold', 'Gini_Gain']].to_string(index=False))

print("\n" + "="*70)
print(f"✓ Highest Gini Gain: {gini_df.iloc[0]['Feature']} (Gain = {gini_df.iloc[0]['Gini_Gain']:.6f})")
print("="*70)
````

## 10. Feature Importance Analysis

### 10.1 Understanding Feature Importance

Feature importance scores indicate how valuable each feature is in constructing the decision tree. In sklearn's DecisionTreeClassifier, importance is calculated based on:

**Gini Importance (Mean Decrease in Impurity)**:
$$Importance(f) = \sum_{nodes\ using\ f} \frac{n_{samples}}{n_{total}} \times \Delta Impurity$$

where:
- The sum is over all nodes that split on feature $f$
- $n_{samples}$ = number of samples at that node
- $n_{total}$ = total number of training samples
- $\Delta Impurity$ = reduction in impurity (Gini or entropy) from the split

**Properties**:
- Scores sum to 1.0
- Higher score = more important feature
- Based on how much each feature contributes to reducing impurity

### 10.2 Extract Feature Importances


````python
# Extract feature importances from the trained model
print("="*70)
print("FEATURE IMPORTANCE SCORES")
print("="*70)

feature_importances = dt_classifier.feature_importances_

# Create DataFrame for better visualization
importance_df = pd.DataFrame({
    'Feature': feature_columns,
    'Importance': feature_importances
}).sort_values('Importance', ascending=False)

print(importance_df.to_string(index=False))

# Verification
print(f"\nSum of importances: {feature_importances.sum():.6f} (should be 1.0)")

print("\n" + "="*70)
print("FEATURE IMPORTANCE INTERPRETATION")
print("="*70)

for idx, row in importance_df.iterrows():
    if row['Importance'] > 0.10:
        level = "Very Important"
        marker = "⭐⭐⭐"
    elif row['Importance'] > 0.05:
        level = "Important"
        marker = "⭐⭐"
    elif row['Importance'] > 0.01:
        level = "Moderately Important"
        marker = "⭐"
    else:
        level = "Low Importance"
        marker = "·"

    print(f"{marker} {row['Feature']:25s}: {row['Importance']:.4f} ({row['Importance']*100:5.2f}%) - {level}")
````

````python
# Visualize feature importances
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Bar plot
colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(importance_df)))
axes[0].barh(importance_df['Feature'], importance_df['Importance'], color=colors, alpha=0.8)
axes[0].set_xlabel('Importance Score', fontweight='bold', fontsize=12)
axes[0].set_title('Feature Importance\n(Based on Mean Decrease in Impurity)',
                 fontweight='bold', fontsize=13)
axes[0].grid(axis='x', alpha=0.3)
axes[0].invert_yaxis()

# Pie chart (top features)
top_n = 5
top_features = importance_df.head(top_n).copy()
other_importance = importance_df.iloc[top_n:]['Importance'].sum()

if other_importance > 0:
    top_features = pd.concat([
        top_features,
        pd.DataFrame({'Feature': ['Others'], 'Importance': [other_importance]})
    ])

axes[1].pie(top_features['Importance'], labels=top_features['Feature'],
           autopct='%1.1f%%', startangle=90, colors=colors)
axes[1].set_title(f'Feature Importance Distribution\n(Top {top_n} Features + Others)',
                 fontweight='bold', fontsize=13)

plt.tight_layout()
plt.show()

print("✓ Feature importance visualization complete")
````
